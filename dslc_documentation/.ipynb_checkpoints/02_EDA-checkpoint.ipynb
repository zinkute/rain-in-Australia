## 1. Temperature

<span style="font-family:Cambria">
There are 4 variables related to temperature - min, max, 9 AM and 3 PM temperature each day for 49 different locations. I also have 10 years of data, therefore it might be difficult to plot every locations through years and see meaningful data.
</span>

```python
# add detailed date columns - year, month
rain_data_clean['Year'] = rain_data_clean['Date'].dt.strftime('%Y')
rain_data_clean['Month'] = rain_data_clean['Date'].dt.strftime('%m')
```

```python
average_temp_9am = rain_data_clean.groupby(["Year"])['Temp9am'].mean().reset_index()
average_temp_3pm = rain_data_clean.groupby(["Year"])['Temp3pm'].mean().reset_index()
average_max_temp = rain_data_clean.groupby(["Year"])['MaxTemp'].mean().reset_index()
average_min_temp = rain_data_clean.groupby(["Year"])['MinTemp'].mean().reset_index()

# create subplots
fig, axes = plt.subplots(1, 4, figsize=(15, 6))

sns.lineplot(data=average_temp_9am, x='Year', y='Temp9am', ax=axes[0])
axes[0].set_title("Average temperature in the morning")
axes[0].set_ylim(0, 45)
axes[0].tick_params(labelrotation=45)

sns.lineplot(data=average_temp_3pm, x='Year', y='Temp3pm', ax=axes[1])
axes[1].set_title("Average temperature in the afternoon")
axes[1].set_ylim(0, 45)
axes[1].tick_params(labelrotation=45)

sns.lineplot(data=average_max_temp, x='Year', y='MaxTemp', ax=axes[2])
axes[2].set_title("Average maximum temperature")
axes[2].set_ylim(0, 45)
axes[2].tick_params(labelrotation=45)

sns.lineplot(data=average_min_temp, x='Year', y='MinTemp', ax=axes[3])
axes[3].set_title("Average minimum temperature")
axes[3].set_ylim(0, 45)
axes[3].tick_params(labelrotation=45)

plt.tight_layout()
plt.show()
```

Then I will plot temperatures for every location in different states in a different plot to see them all together. This will help to visualise how weather in different parts of Australia looks like.

```python
# temperature data for Tasmania weather stations
tasmania_data = rain_data_clean.query("State == 'Tasmania'")

# create subplots
fig, axes = plt.subplots(4, 1, figsize=(12, 15))

sns.boxplot(data=tasmania_data, x='Year', y='Temp9am', hue = 'Location', ax=axes[0])
axes[0].set_title("Temperature at 9 AM in Tasmania weather stations")
sns.move_legend(axes[0], "upper left", bbox_to_anchor=(1, 1))

sns.boxplot(data=tasmania_data, x='Year', y='Temp3pm', hue = 'Location', ax=axes[1])
axes[1].set_title("Temperature at 3 PM in Tasmania weather stations")
sns.move_legend(axes[1], "upper left", bbox_to_anchor=(1, 1))

sns.boxplot(data=tasmania_data, x='Year', y='MaxTemp', hue = 'Location', ax=axes[2])
axes[2].set_title("Max temperature in Tasmania weather stations")
sns.move_legend(axes[2], "upper left", bbox_to_anchor=(1, 1))

sns.boxplot(data=tasmania_data, x='Year', y='MinTemp', hue = 'Location', ax=axes[3])
axes[3].set_title("Min temperature in Tasmania weather stations")
sns.move_legend(axes[3], "upper left", bbox_to_anchor=(1, 1))

plt.tight_layout()  # Avoid overlap

plt.show()
```

I will check how does months look like. Maybe we have data from June onwards, or is it missing for whole year?

```python
# checking 2007 temperature data by month
missing_temp_by_month_2007 = rain_data_clean[["Year", "Month", "MinTemp", "MaxTemp", "Temp9am", "Temp3pm"]].query("Year == '2007'").groupby(['Month']).count()
```

```python
# checking 2008 temperature data by month
missing_temp_by_month_2008 = rain_data_clean[["Year", "Month", "MinTemp", "MaxTemp", "Temp9am", "Temp3pm"]].query("Year == '2008'").groupby(['Month']).count()
```

```python
# checking 2009 temperature data by month (to compare)
missing_temp_by_month_2009 = rain_data_clean[["Year", "Month", "MinTemp", "MaxTemp", "Temp9am", "Temp3pm"]].query("Year == '2009'").groupby(['Month']).count()
```

I will visualise the findings above to get a better understanding how does 2008 and 2007 records look like, and compare them to 2009 data, which looks OK.

```python
# create subplots
fig, axes = plt.subplots(3, 1, figsize=(12, 8))

sns.barplot(data=missing_temp_by_month_2007, x='Month', y='MinTemp', order=list(range(1, 13)), ax=axes[0])
axes[0].set_title("Min Temperature record count by month in 2007")
axes[0].set_ylim(0, 1500)

sns.barplot(data=missing_temp_by_month_2008, x='Month', y='MinTemp', ax=axes[1])
axes[1].set_title("Min Temperature record count by month in 2008")
axes[1].set_ylim(0, 1500)

sns.barplot(data=missing_temp_by_month_2009, x='Month', y='MinTemp', ax=axes[2])
axes[2].set_title("Min Temperature record count by month in 2009")
axes[2].set_ylim(0, 1500)

plt.tight_layout()  # Avoid overlap

plt.show()
```

It is clearly visible how many records are missing in 2007 and 2008 in comparison to 2009 - I will not visualise remaining variables as numbers looks very similar. But I will keep that in mind - I might need to remove 2007 and 2008 data from my investigation at all.

## 2. Rainfall
</span>

I'll try to check mean values for Rainfall data in each weather station. I will try to compare average Rainfall each year, to see if there is something interesting to see.

```python
average_rainfall = rain_data_clean.groupby(["Year"])['Rainfall'].mean().reset_index()

sns.lineplot(data=average_rainfall, x='Year', y='Rainfall')

plt.show()
```

```python
average_yearly_rainfall = rain_data_clean.groupby(["Year", "Location"])['Rainfall'].mean().reset_index()

palette = sns.color_palette("tab20", n_colors=49)  # Example with a base palette
plt.figure(figsize=(15, 10))  # Width = 10 inches, Height = 6 inches
ax = sns.lineplot(data=average_yearly_rainfall, x='Year', y='Rainfall', hue='Location', errorbar=None, palette=palette, style='Location')

sns.move_legend(ax, "upper left", bbox_to_anchor=(1, 1))
plt.show()
```

From the above visualisation, I can see that during the years highest average rainfall was recorded in Cairns, Darwin and Coffs Harbour.

```python
highlight_locations = ['Cairns', 'Darwin', 'Coffs Harbour']

all_locations = average_yearly_rainfall['Location'].unique()

# Create colors for the highlighted ones
highlight_colors = sns.color_palette('Set1', n_colors=len(highlight_locations))

# Create a mapping from location to color
palette = {}
color_idx = 0
for loc in all_locations:
    if loc in highlight_locations:
        palette[loc] = highlight_colors[color_idx]
        color_idx += 1
    else:
        palette[loc] = 'lightgrey'

# Plot
plt.figure(figsize=(15, 10))
ax = sns.lineplot(
    data=average_yearly_rainfall,
    x='Year',
    y='Rainfall',
    hue='Location',
    palette=palette,
    errorbar=None
)

sns.move_legend(ax, "upper left", bbox_to_anchor=(1, 1))
plt.show()
```

I will also check how many records are missing for Rainfall data as well. Is it the same as for Temperature, where majority of records are missing for 2007-2008?

```python
# checking 2007 rainfall data by month
missing_rainfall_by_month_2007 = rain_data_clean[["Year", "Month", "Rainfall"]].query("Year == '2007'").groupby(['Month']).count()

# checking 2008 rainfall data by month
missing_rainfall_by_month_2008 = rain_data_clean[["Year", "Month", "Rainfall"]].query("Year == '2008'").groupby(['Month']).count()

# checking 2009 rainfall data by month (to compare)
missing_rainfall_by_month_2009 = rain_data_clean[["Year", "Month", "Rainfall"]].query("Year == '2009'").groupby(['Month']).count()

# create subplots
fig, axes = plt.subplots(3, 1, figsize=(12, 8))

sns.barplot(data=missing_rainfall_by_month_2007, x='Month', y='Rainfall', order=list(range(1, 13)), ax=axes[0])
axes[0].set_title("Rainfall record count by month in 2007")
axes[0].set_ylim(0, 1500)

sns.barplot(data=missing_rainfall_by_month_2008, x='Month', y='Rainfall', ax=axes[1])
axes[1].set_title("Rainfall record count by month in 2008")
axes[1].set_ylim(0, 1500)

sns.barplot(data=missing_rainfall_by_month_2009, x='Month', y='Rainfall', ax=axes[2])
axes[2].set_title("Rainfall record count by month in 2009")
axes[2].set_ylim(0, 1500)

plt.tight_layout()  # Avoid overlap

plt.show()
```

It is definitely the same - I will have to think about it later on when creating a model. Does such missing data numbers impact my model? Maybe it is better to ignore years with no data and focus on 2009 onwards.

Owerall, rainfall data doesn't seem incorrect. There are no strange mean in any of the fields, like it was visible with temperature in Melbourne.

## 3. Wind
</span>

There are few wind-related columns in the data set. There are two different types of wind data that we have: wind and wind gust. Both variables have wind direction and speed information. At this point I will check for outliers and overall trends; maybe wind speed changes based on season? I will pay attention to missing data as well. I will examine wind data per state too.

```python
average_wind_speed_9am = rain_data_clean.groupby(["Year", "Location"])['WindSpeed9am'].mean().reset_index()

palette = sns.color_palette("tab20", n_colors=49)  # Example with a base palette
plt.figure(figsize=(15, 10))  # Width = 10 inches, Height = 6 inches
ax = sns.lineplot(data=average_wind_speed_9am, x='Year', y='WindSpeed9am', hue='Location', errorbar=None, palette=palette, style='Location')

sns.move_legend(ax, "upper left", bbox_to_anchor=(1, 1))
plt.show()
```

Certainly there are locations where average wind speed is way higher than other locations.

```python
average_wind_speed_3pm = rain_data_clean.groupby(["Year", "Location"])['WindSpeed3pm'].mean().reset_index()

palette = sns.color_palette("tab20", n_colors=49)  # Example with a base palette
plt.figure(figsize=(15, 10))  # Width = 10 inches, Height = 6 inches
ax = sns.lineplot(data=average_wind_speed_3pm, x='Year', y='WindSpeed3pm', hue='Location', errorbar=None, palette=palette, style='Location')

sns.move_legend(ax, "upper left", bbox_to_anchor=(1, 1))
plt.show()
```

```python
average_wind_speed_9am = rain_data_clean.groupby(["Year"])['WindSpeed9am'].mean().reset_index()
average_wind_speed_3pm = rain_data_clean.groupby(["Year"])['WindSpeed3pm'].mean().reset_index()
average_wind_speed_gust = rain_data_clean.groupby(["Year"])['WindGustSpeed'].mean().reset_index()

# create subplots
fig, axes = plt.subplots(1, 3, figsize=(15, 6))

sns.lineplot(data=average_wind_speed_9am, x='Year', y='WindSpeed9am', ax=axes[0])
axes[0].set_title("Average Wind Speed in the morning")
axes[0].set_ylim(0, 45)
axes[0].tick_params(labelrotation=45)

sns.lineplot(data=average_wind_speed_3pm, x='Year', y='WindSpeed3pm', ax=axes[1])
axes[1].set_title("Average Wind Speed in the afternoon")
axes[1].set_ylim(0, 45)
axes[1].tick_params(labelrotation=45)

sns.lineplot(data=average_wind_speed_gust, x='Year', y='WindGustSpeed', ax=axes[2])
axes[2].set_title("Average Wind Gust Speed")
axes[2].set_ylim(0, 45)
axes[2].tick_params(labelrotation=45)

plt.show()
```

It doesn't look like there are any anomalies with wind data, just by looking into average speed in both wind and wind gust variables both looks OK. Wind gusts are expected to be way higher than average speed. Afternoon wind tends to be a little higher than in the morning. Let's see how wind looks in different states.
